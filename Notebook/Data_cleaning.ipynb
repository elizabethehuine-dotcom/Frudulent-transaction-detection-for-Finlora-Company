SUMMARY/OBSERVATION

1. Dataset Size: The dataset has 11,400 rows and 26 columns. All column names are formatted correctly.

2. Data Type Fixes Needed: timestamp needs to be changed from text to a date/time format. amount_sro needs to be changed from text to a number so we can do math with it.

3. Missing Data: Several columns have missing rows that we need to clean. timestamp is missing 29 values. amount_usd and ip_address are missing 305 values. fee and device_trust_score are missing 295 values. ip_country is missing 301 values, and kyc_tier is missing 300 values.

We must use special training techniques to fix this.

DATA CLEANING AND PREPARATION

```python
#Converting data types and timestamp to datetime format
customer_transaction_data['timestamp'] = pd.to_datetime(customer_transaction_data['timestamp'], errors='coerce')

#Converting data type of income usd to float
customer_transaction_data['amount_src'] = pd.to_numeric(customer_transaction_data['amount_src'], errors='coerce')
```

```python
#checking the dataset again to see if the changes have been made
customer_transaction_data.info()
```

Data Type Conversion: We converted timestamp to actual dates and amount_src to numbers using Pandas (errors='coerce').

Hidden Errors Found: This conversion caught corrupted or badly formatted text that wasn't noticed before. Because the computer couldn't read them, it turned them into missing values (NaN).

Increase in Missing Data: Missing rows in timestamp jumped from 29 to 61.Missing rows in amount_src went from 0 to 4.

Next Steps: These broken records were safely cleaned up in the next stage of data cleaning.

```python
#Rechecking for missing values after data type conversion
customer_transaction_data.isnull().sum()
```

A. Variables to Fill In (Imputation)

1. amount_usd (305 missing): We will calculate the missing values by multiplying amount_src by the currency exchange rate.

2. fee (295 missing): We will fill these using the median fee of the specific transaction channel. Any leftover missing fees will get the overall median fee.

3. ip_country (301 missing): We will fill these by copying over the user's home_country.

4. kyc_tier (300 missing): We will fill these with the most common tier (the mode).

5. device_trust_score (295 missing): We will fill these using the median score of users with the same kyc_tier and new_device status. Any leftovers will get the overall median score.

B. Variables to Delete (Dropping Rows)

1. timestamp (61 missing) & amount_src (4 missing): We will delete these rows using .dropna(). The number of missing rows is too small to matter.

2. ip_address (305 missing): We will delete these rows because IP addresses are completely unique and cannot be guessed or filled in accurately.

```python
#calculating exchange rates per currency
exchange_rates = customer_transaction_data[customer_transaction_data['amount_usd'].notna()].groupby('source_currency').apply(lambda x: (x['amount_usd'] / x['amount_src'].mean()).mean()
).to_dict()
exchange_rates
```

```python
#filling missing values in the amount_usd column using the calculated exchange rates
customer_transaction_data['amount_usd'] = customer_transaction_data.apply(lambda row: row['amount_usd'] if not pd.isna(row['amount_usd']) else row['amount_src'] * exchange_rates.get(row['source_currency'], 1), axis=1)
```

```python
#filling the missing values of the fee variable
if 'fee' in customer_transaction_data.columns:
    customer_transaction_data['fee'] = customer_transaction_data.groupby('channel')['fee'].transform(lambda y: y.fillna(y.median()))
customer_transaction_data['fee'] = customer_transaction_data['fee'].fillna(customer_transaction_data['fee'].median())
```

```python
#filling the missing of ip_country variable
if ('ip_country' in customer_transaction_data.columns):
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])
```

```python
#filling the missing values of device_trust_score varaible
if 'device_trust_score' in customer_transaction_data.columns:
    if{'new_device', 'kyc_tier'}.issubset(customer_transaction_data.columns):
        customer_transaction_data['device_trust_score'] = customer_transaction_data.groupby(['new_device', 'kyc_tier'])['device_trust_score'].transform(lambda x: x.fillna(x.median()))
    customer_transaction_data['device_trust_score'] = customer_transaction_data['device_trust_score'].fillna(customer_transaction_data['device_trust_score'].median())
 
```

```python
# dropping the timestamp, amount_src, and ip_address missing values columns as they are not needed for the analysis
customer_transaction_data.dropna(subset=['timestamp', 'amount_src', 'ip_address'], inplace=True)
```

```python
# rechecking for missing values after data cleaning
customer_transaction_data.isnull().sum()
```
